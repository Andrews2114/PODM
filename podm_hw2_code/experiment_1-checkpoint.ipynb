```python
import numpy as np
import matplotlib.pyplot as plt

from bandit_environment import BernoulliBandit
from bandit_algorithm import ExploreThenCommit, EpsilonGreedy
from plotting import finalize_plot
```

### Explore-Then-Commit

```python
# parameters for bandit
# number of arms
n_arms = 2
# suboptimality gap
delta = 0.2
prob_vec = [0.5 - delta / 2, 0.5 + delta / 2]
# time horizon
time_horizon = 2000

# parameters for explore-then-commit
explore_time_etc = 100
seed_etc = 0  # to fix randomness

# parameters for epsilon-greedy
eps = 0.1
seed_e_greedy = 0  # to fix randomness
# try seed 4195
```

```python
# initialise environment
env1 = BernoulliBandit(probabilities=prob_vec, random_seed=seed_etc)
# initialise bandit algorithm
etc_alg = ExploreThenCommit(
    n_arms=n_arms, T_explore=explore_time_etc, random_seed=seed_etc
)
```

```python
# set up loop
regret = np.zeros(time_horizon)
for timestep in range(time_horizon):
    # algorithm chooses an arm
    chosen_arm = etc_alg.select_arm()
    # environment provides a reward
    reward = env1.pull_arm(chosen_arm)
    # algorithm updates its internal state
    etc_alg.update(chosen_arm, reward)
    # compute regret
    instant_regret = env1.get_regret(chosen_arm)
    if timestep == 0:
        regret[timestep] = instant_regret
    else:
        regret[timestep] = regret[timestep - 1] + instant_regret
```

```python
# plot regret
plt.plot(range(time_horizon), regret, label="ETC")
# added dotted vertical line to indicate end of exploration phase
plt.axvline(
    x=n_arms * explore_time_etc, color="r", linestyle="--", label="End of Exploration"
)
finalize_plot(xlabel="Time step", ylabel="Cumulative Regret", title="Cumulative Regret (ETC)")
```

```python
# plot chosen arm
chosen_arms = np.array(etc_alg.arm_history)
plt.scatter(range(time_horizon), chosen_arms, label="Chosen Arm", s=1)
plt.axvline(
    x=n_arms * explore_time_etc, color="r", linestyle="--", label="End of Exploration"
)
plt.yticks(np.arange(n_arms))
finalize_plot(xlabel="Time step", ylabel="Chosen Arm", title="Chosen Arm over Time (ETC)")
```

```python
# plot arm estimates over time
for i in range(n_arms):
    plt.plot(range(time_horizon),
             etc_alg.get_arms_estimates_history()[:, i],
             label="Arm" + str(i))
plt.axvline(
    x=n_arms * explore_time_etc, color="r", linestyle="--", label="End of Exploration"
)
finalize_plot(xlabel="Time step", ylabel="Arm reward estimate",title="Reward estimate over time (ETC)")
```

```python
# Generate 1e8 Gaussian samples N(0, 1)
samples = 1e8
gaussian_samples = np.random.normal(loc=0.0, scale=1.0, size=int(samples))

# Generate 1e8 Student's t samples with df = 3 and variance 1
df = 3
t_scale = np.sqrt((df - 2) / df)  # since Var[t_df] = df / (df - 2) = 3
t_samples = np.random.standard_t(df=df, size=int(samples)) * t_scale

# Optional diagnostics
print("Gaussian mean/var:", gaussian_samples.mean(), gaussian_samples.var())
print("t mean/var:", t_samples.mean(), t_samples.var())
```

```python
deltas = (1e-4, 1e-3, 1e-2, 1e-1)

print(f"{'delta':>10} {'threshold':>12} {'Gaussian tail':>16} {'t tail':>16} {'Gauss holds?':>14} {'t holds?':>10}")

for delta_value in deltas:
    threshold = np.sqrt(2 * np.log(2 / delta_value))
    gaussian_tail = np.count_nonzero(np.abs(gaussian_samples) >= threshold) / gaussian_samples.size
    t_tail = np.count_nonzero(np.abs(t_samples) >= threshold) / t_samples.size
    
    gaussian_holds = gaussian_tail <= delta_value
    t_holds = t_tail <= delta_value
    
    print(
        f"{delta_value:10.1e} "
        f"{threshold:12.4f} "
        f"{gaussian_tail:16.6g} "
        f"{t_tail:16.6g} "
        f"{str(gaussian_holds):>14} "
        f"{str(t_holds):>10}"
    )
```

```python
import numpy as np
import matplotlib.pyplot as plt

from bandit_environment import GaussianBandit, UniformBandit, StudentTBandit
from bandit_algorithm import UCB, UCB_Var

means_dist = np.array([1.0, 1.0, 2.0])
# Use unequal known variances so UCB_Var differs meaningfully from standard UCB.
variances_dist = np.array([0.5, 1, 1])
horizon_dist = 3000
runs_dist = 50
error_bound_dist = 0.01

student_dfs_dist = np.full(len(means_dist), 5.0)
student_scales_dist = np.sqrt(
    variances_dist * (student_dfs_dist - 2) / student_dfs_dist
)
uniform_intervals_dist = [
    (
        mean - np.sqrt(3 * variance),
        mean + np.sqrt(3 * variance),
    )
    for mean, variance in zip(means_dist, variances_dist)
]

# Each environment uses its own constructor parameters while preserving the
# same arm means and variances across distribution families.
bandit_factories_dist = {
    "Gaussian": lambda seed: GaussianBandit(
        means_dist, variances_dist, random_seed=seed
    ),
    "Uniform": lambda seed: UniformBandit(
        uniform_intervals_dist, random_seed=seed
    ),
    "Student-t": lambda seed: StudentTBandit(
        student_dfs_dist,
        means_dist,
        student_scales_dist,
        random_seed=seed,
    ),
}
algorithm_factories_dist = {
    "UCB": lambda seed: UCB(
        n_arms=len(means_dist),
        error_prob_bound=error_bound_dist,
        random_seed=seed,
    ),
    "UCB_Var": lambda seed: UCB_Var(
        n_arms=len(means_dist),
        error_prob_bound=error_bound_dist,
        variances_vec=variances_dist,
        random_seed=seed,
    ),
}

results_dist = {
    (bandit_name, algorithm_name): np.zeros((runs_dist, horizon_dist))
    for bandit_name in bandit_factories_dist
    for algorithm_name in algorithm_factories_dist
}

for run_dist in range(runs_dist):
    seed_dist = 1000 + run_dist

    for bandit_name, bandit_factory in bandit_factories_dist.items():
        for algorithm_name, algorithm_factory in algorithm_factories_dist.items():
            env_dist = bandit_factory(seed_dist)
            algorithm_dist = algorithm_factory(seed_dist)
            cumulative_regret_dist = 0.0

            for timestep_dist in range(horizon_dist):
                arm_dist = algorithm_dist.select_arm()
                reward_dist = env_dist.pull_arm(arm_dist)
                algorithm_dist.update(arm_dist, reward_dist)
                cumulative_regret_dist += env_dist.get_regret(arm_dist)
                results_dist[(bandit_name, algorithm_name)][
                    run_dist, timestep_dist
                ] = cumulative_regret_dist

fig, axes = plt.subplots(
    1, len(bandit_factories_dist), figsize=(16, 4), sharey=True, squeeze=False
)
for ax, bandit_name in zip(axes[0], bandit_factories_dist):
    for algorithm_name in algorithm_factories_dist:
        mean_regret_dist = results_dist[
            (bandit_name, algorithm_name)
        ].mean(axis=0)
        ax.plot(
            np.arange(1, horizon_dist + 1),
            mean_regret_dist,
            label=algorithm_name,
        )

    ax.set_title(f"{bandit_name} rewards")
    ax.set_xlabel("Iteration")
    ax.grid(True, alpha=0.3)
    ax.legend()

axes[0, 0].set_ylabel(f"Mean cumulative regret ({runs_dist} runs)")
fig.suptitle("UCB vs. UCB_Var across bandit reward distributions")
fig.tight_layout(rect=(0, 0, 1, 0.92))
plt.show()

for bandit_name in bandit_factories_dist:
    ucb_final_regret = results_dist[(bandit_name, "UCB")][:, -1].mean()
    ucb_var_final_regret = results_dist[(bandit_name, "UCB_Var")][:, -1].mean()
    print(
        f"{bandit_name}: mean final regret: "
        f"UCB={ucb_final_regret:.2f}, UCB_Var={ucb_var_final_regret:.2f}"
    )
```
